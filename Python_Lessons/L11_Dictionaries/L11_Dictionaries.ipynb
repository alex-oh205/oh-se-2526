# L11:  Dictionaries

```python
elements = {
    "hydrogen": 1,
    "helium": 2,
    "lithium": 3,
    "beryllium": 4,
}
```

## Adding and Replacing Values

```python

elements["carbon"] = 6
```

## Accessing Values

```python
elements["carbon"] = 6
print(elements["carbon"])
```

## Removing Keys

```python
elements.pop("helium", None)
print(elements.get("helium", None))
```

## Traversing a Dictionary

```python
for element in elements:
    print(element, elements[element])
```

## Ordering Dictionary Entries by Key

```python
orderedKeys = list(elements.keys())
print(orderedKeys)
orderedKeys.sort()
print(orderedKeys)
```
